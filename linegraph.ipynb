```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[5,6,8,10,15]
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.xlim(1,10)
plt.ylim(5,15)
plt.title("LINE PLOT")
plt.plot(x,y)
plt.savefig("sample.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[40,50,60,80,100]
x1=[1,2,3,4]
y1=[55,65,85,95]
plt.xlabel("roll no")
plt.ylabel("marks")
plt.xlim(1,6)
plt.ylim(40,110)
plt.title("STUDENTS DETAILS")
plt.plot(x,y,label="Line1",color='purple',linestyle='dashdot',linewidth=3,marker='v',markerfacecolor='blue',markersize=12)
plt.plot(x1,y1,label="Line-2",alpha=0.5,color='yellow',linestyle='dashed',linewidth=5,marker='D',markerfacecolor='green',markersize=13)
plt.legend()
plt.savefig("sample1.png")
plt.show()
```
